## Map function 

> - When each item is transformed we say that the operation is a **mapping**, or just a **map** of the original list. 
>
> - When some items are omitted, we call it a **filter**.
>
> - Python provides built-in functions **map** and **filter**
>
> - **map takes two arguments**, a **function** and a **sequence**. 
>
> - The function is the mapper that transforms items. It is automatically applied to each item in the sequence
>
> - The map function produces an iterator and so whenever we do that to a list we have **map object** as our return type so we need to convert it to a tuple or list using casting 
>
> - Once a map function is used it can not be reused again unless we make a new one 
>
>>  #### **Syntax**
>> **map(function, iterable)**
>> - function: The function to apply to every element of the iterable.
>> - iterable: One or more iterable objects (list, tuple, etc.) whose elements will be processed.
> 
> ### **Why do we use map() function ?**
> 
> The main reason behind using map function is **memory effeciency**, we can think of this function as an on-demand live stream. The function takes the **current requested** functioin, the iterable, current position and perform the procedure without having python to store any additional function either built in or user defined.
>
> Instead of building huge amount of memory storage to store the reuslts python only stores the function, the iterable, and the pointer instead of storing the acutal values as to what lists do 

```python
s = [1,2,3,4]
result = map((lambda x: x+2),s)
print(f"printing the map without casting: {result}")
print(f"printing the map without casting: {list(result)}")
```

```python
s = ['1', '2', '3', '4']
res = map(int, s)
print(list(res))
```

## Filter 

> It is a built in python function **filter** which perform a procedure of keeping selected items while excluding the unwanted noes 
>
> Filter functions by automatically performing function for each item without the need for an accumulator  
>
>>  #### **Syntax**
>> **filter(function, iterable)**
>> - function: The function to apply to every element of the iterable **Function MUST be of conditional statement type**.
>> - iterable: **ONLY One** iterable objects (list, tuple, etc.) whose elements will be processed.
> 
> ### **Why do we use filter() function and how does it differ form map() ?**
> 
> The main different feature between **filter** and **map** function is that : 
> 
> - map() retreives an iterable map object, and depending on the function given wheter it is conditional or non-conditional it returns all the value without excluding any of the items 
>
> - filter() function retreives a filter object, and only depends on **conditional statement function** 

#### Try using map() function with the concept of conditional statement (filter())

```python
s = [1,2,3,4]
result = map((lambda x: x+2 == 4),s)
print(f"printing the map without casting: {result}")
print(f"printing the map without casting: {list(result)}")
```

#### Try using filter() function without the concept of conditional statement (map())

```python
s = [1,2,3,4]
result = filter((lambda x: x+2),s) 
print(f"printing the filter without casting: {result}")
print(f"printing the filter without conditional statement: {list(result)}")
```

#### Try using filter() function with the concept of conditional statement 

```python
s = [1,2,3,4]
result = filter((lambda x: x+2 == 4 ),s) # if true then store the value else exclude it 
print(f"printing the filter without casting: {result}")
print(f"printing the filter with conditional statement: {list(result)}")
```

## List Comprehension  

> An alternative way to perform map and filter operation and a way to create a list from an old list this is called **List Comprehension*
>
>>  #### **Syntax**
>> **[<transformer_expression> for <loop_var> in <sequence> if <filtration_expression>]**

? 2. The for loop below produces a list of numbers greater than 10. Below the given code, use list comprehension to accomplish the same thing. Assign it the the variable lst2. Only one line of code is needed.

```python
L = [12, 34, 21, 4, 6, 9, 42]
lst = []
for x in L:
    if x > 10:
        lst.append(x)
print(lst)

lst2 = [lst2.append(i) for i in L if i>10] 
```

## Zip

> **zip()** is a built-in function that returns a zip object, which is an iterator of **tuples** where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together etc.
>
>>  #### **Syntax**
>> **zip(iterable1,iterable2)** 
>
> **Outputs a tuple**

```python
l1 = [1,2,3,5,6]
l2 = [1,2,3,4,5]
print(list(zip(l1,l2)))
```
