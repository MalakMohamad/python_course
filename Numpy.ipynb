# What is Numpy ?

> Numpy is a python library is the fundamental library for the numerical computation. It is an integral part of the Scientific Python Ecosystem.

> We make use of numpy library to store raw data. It has a basic
yet very versatile data structure known as Ndarray which refers to **N-dimensional array**

> Designed to handle large scientific, mathematical, and multidimensional datasets (vectors, matrices, tensors).

# But why do we need to use Numpy when we can have list or dict or even set data strucutre to perfrom the mentioned tasks ?

> The reason behind using Numpy library is due to its performance speed when handling large data set as it performs calculations at native C-speed using vectorization (operating on whole arrays without slow Python for loops), serving as the backbone for Pandas, Scikit-learn, and PyTorch.

> Additionally, unlike Python lists, all items in an ndarray must share the exact same data type.

> To use numpy library in python we must first import it as shown:  
 `import numpy as np` where np is an alias for numpy that is commonly used among developers  

````python
import numpy as np
````

> NumPy arrays can be defined using Python sequences such as lists and tuples. Both list and tuples define the ndarray creation:
>
>1 - a list of numbers forms a 1D array
>
>2 - a list of lists form a 2D array
>
> 3- Nested lists or tuples will form a higher array dimension as in N-dimensional array  
>
> Since any numerical data in the form of lists or tuples they must be converted to array inorder to have them be used in NumPy.
>
> **As NumPy base data type is array object.**

>
# How can we define a new array using NumPy ?
> To define a new array using NumPy we must first **import numpy** library to get access to all of its methods
>
> numpy.array() is used to convert lists into array
>
> Then we make use of the library name or its alias (commonly used is **np** so it becomes **import nmupy as np**) and the since we are dealing with built in method we make us of dot notation concept to invoke the **array(paramter: list or tuple)** method


```
numpy.array(argument)
```
>
> The following syntax will illustrate the type casting of lists and tuples to make them useful and usable for NumPy library

````python
# creating a list to illustrate type conversion
lst1 = [1,2]
lst2 = [3,5]
print(f"first list is of type: {type(lst1)}\nholding the data: {lst1}\nsecond list is of type {type(lst2)}\nholding the data: {lst2}\n")

######## creating a 1D array using lst1
d1 = np.array(lst1)
print(f"feeding the list one to array of d1: {d1}\nis of type: {type(d1)} and it is a 1D array\n")

######## creating a 1D array using lst2
d1_2 = np.array(lst2)
print(f"feeding the list two to array of d1: {d1}\nis of type: {type(d1)} and it is a 1D array\n")

######## creating a 2D array using both lst1 and lst2
d2 = np.array([lst1,lst2])
print(f"feeding the list one and two to array of d2:\n{d2}\nis of type: {type(d2)} and it is a 2D array\n")
````

## numpy.zeros()
>>### **Syntax**
>> - **numpy.zeros((colNum, rowNum))**

````python
a2 = np.zeros((3,3))
print(f"This is array with all zeros in it: \n{a2}\n")
````

## numpy.ones()
>> ### **Syntax**
>> - **numpy.ones((colNum, rowNum))**

````python
a1 = np.ones((3,3))
print(f"This is array with all ones in it: \n{a1}")
````

## numpy.arange()
>> ### **Syntax**
>> - **numpy.arange(statNum,stopNum,jumps)** 
>>
>> or
>>
>> - **numpy.arange(num)**
>>
>> or
>>
>> - **numpy.arange(statNum,stopNum)** 
>>

````python
## usign start,end and number of jumps 
a3 = np.arange(0,10,2)
print(f"This is array with values from 0 to 10 skipping 2 : \n{a3}\n")

## using single number
a4 = np.arange(16)
print(f"This is array with values from 0 to 15 inclusive : \n{a4}\n")

## using start and end only 
a5 = np .arange(-1,21)
print(f"This is array with values from -1 to 20 inclusive : \n{a5}\n")

````

## numpy.linespace()
>> ### **Syntax**
>> - **numpy.linspace(startNum, stopNum, jumps)** 
>>
>> Creates array with specified number of elements and spaced equally between specified begining to end 
value. The calcualtion that happens behind this function is an Arithmetic Sequence formula:
>> 
>> **(endNum-startNum)/(steps-1)** 


````python
startNum = 1
endNum = 4
steps = 3
a6 = np.linspace(startNum,endNum,steps)
print(f"{a6}")
print(f"Using the arithmetic formula to calculate the interval spaces: {(endNum-startNum)/(steps-1)}")
## testing case 
test = (endNum-startNum)/(steps-1)
assert startNum + test == 2.5, "test case failed"
````
