## What are decorators in python

> a convenient shorthand for reassigning a function name to the output of a higher-order function.
>
> so it informs python how a function must act when executed 
> 
> We have a type of decorator called property decorator 
> it defines a set to a method to make it act as getter only. The effect of having a built in setter to set a specific variable to a value is to use **decorators** with syntax given: 
`@ varName.setter` varName is the variable that we want to set it to a new value

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name   # getter

    @name.setter
    def name(self, value):
        self._name = value  # setter
p = Person("malak")
print(p.name)
p.name = "Sara"
print(p.name)
```

## Static Method 

> As shown previously that any method defined with parameter **self** is known as instance method 
> Static method does not belong to any object and they so they are also known as class methods. However they can not access instance or class attributes yet they belongs to the class namespace

*  Static method does not require a parameter like class and instance method
*   Static method does not dependent on any particular instance of the class rather it belongs to the clas itself
*  it is a standalone metohd that can not change class attribute or any instance atrribute levels
*   It’s just a utility function that take some parameters and work upon those parameters
*  We can think of it as a validation function

### How can we define a static method in python ? 
 
 > To define static method we maek use of **decorator** **@staticmethod** before method definition 






Goal: Build a utility class for temperature conversions.
- Requirements:Create a class named TemperatureConverter.
- Add a static method celsius_to_fahrenheit(c) that returns $(C * 9/5) + 32$.
- Add a static method fahrenheit_to_celsius(f) that returns $(F - 32) * 5/9$.
- Demonstrate calling both methods directly on the class (without instantiating it).

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(c):
        return f"{(c*(9/5))+32}"
    @staticmethod
    def fahrenheit_to_celsius(f):
        return f"{f-32*(5/9)}"
print(TemperatureConverter.fahrenheit_to_celsius(0))
```

# Class method 

> a class method is a built-in method in python that is defined under a special decorator `@classmethod` defining that the method is related to a class and to an instance.
>
> The metohd receives the `class (cls)` as its first argument rather than `instance (self)`
>
> To deal with class attributes we make use of the class method. Hence they are useful when working iwth the class attributes them self and not to deal with object initialization


```python
class Employee:
  empolyee_count = 0
  company_name = "TechCrop"
  def __init__(self):
    # to access and use the class attribute we call it using class name
    Employee.empolyee_count += 1
  # we must use decorator (@classmethod) to deinf a class method
  @classmethod
  # and we must use cls keyword as first parameter for the class method instead of self keyword
  def display_employee_count(cls):
    return f"Total number of employee is {cls.empolyee_count}"
  @classmethod
  def change_company_name(cls, company_name):
    cls.company_name = company_name

```

# Special operators that can be redefined

> In python operatorsl ike *, /, +, - can be redefined upon overridding their method and changing their original functionality 
>
> __add__(self,other) -> self+other 

>__sub__(self, other) -> self - other

>__mul__(self, other) -> self * other

>__truediv__(self, other) -> self / other

>__floordiv__(self, other) -> self // other

>__mod__(self, other) -> self % other

>__pow__(self, other) -> self ** other

> and many more operators can be redefined through their methods 


```python
class Special:
    def __init__(self, val):
        self.val = val

    def __add__(self, other):
        # Subtracts instead of adding
        return self.val - other

a = Special(2)
print(a + 3) 
```

# Inheritance 

> Inheritance in python OOP that allows a class (called a child or derived class) to inherit attributes and methods from another class (called a parent or base class).\


```python
# Parent Class: Animal
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal name:", self.name)

# Child Class: Dog
class Dog(Animal):
    def __init__(self, name, breed):
        # Calls constructor 
        super().__init__(name)  
        self.breed = breed

    def details(self):
        print(self.name, "is a", self.breed)

d = Dog("Buddy", "Golden Retriever")
d.info()      
d.details()   
```

 ### Method Overriding in Inheritance
> Method overriding allows a child class to provide its own implementation of a method that already exists in the parent class. This enables customized behavior while still maintaining the inheritance relationship.






```python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```

> The above code snippet defines method overriding where Dop class, the method sound form animal is overriden in Dog class
